## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from collections import Counter

plt.rcParams.update({
    'figure.facecolor': '#F8F9FA',
    'axes.facecolor':   'white',
    'axes.spines.top':  False,
    'axes.spines.right':False,
    'axes.grid':        True,
    'grid.alpha':       0.3,
    'font.family':      'sans-serif',
})

PALETTE = ['#4e79a7','#f28e2b','#e15759','#59a14f','#b07aa1','#76b7b2','#edc948','#ff9da7']
COLS_DIM = {
    'Educacion':   '#4e79a7',
    'Vivienda':    '#59a14f',
    'Composicion': '#b07aa1',
    'Ingresos':    '#e15759',
}

print("✅ Librerías OK")
```

## 2. Cargar datos

```python
df = pd.read_csv("dataset_ml_losrios_clean.csv")

df.head(3)
```

## 3. Construir índices IPM (0 = sin carencia · 1 = máxima carencia)

```python

# Paso 1: Normalizar cada variable a 0-1 con MinMaxScaler
# Paso 2: Invertir las que "mayor = mejor"
# Paso 3: Promediar → índice por dimensión

mms = MinMaxScaler()

def construir_indice(df_src, cols, invertir=None):
    cols_ok = [c for c in cols if c in df_src.columns]
    sub = df_src[cols_ok].copy()
    for c in cols_ok:
        sub[c] = sub[c].fillna(sub[c].median())
    scaled = pd.DataFrame(mms.fit_transform(sub), columns=cols_ok, index=df_src.index)
    if invertir:
        for c in [c for c in invertir if c in scaled.columns]:
            scaled[c] = 1 - scaled[c]
    return scaled.mean(axis=1)

# Solo hogares con target válido
df_cl = df[df["pobreza_multi"].notna()].copy().reset_index(drop=True)

# ── EDUCACIÓN ───────────────────────────────────────────────────
df_cl["idx_educacion"] = construir_indice(df_cl,
    cols=["edu_nivel_jefe","edu_nivel_prom","edu_nivel_min",
          "prop_alfabetos","prop_asiste_edu","alfabeto_jefe"],
    invertir=["edu_nivel_jefe","edu_nivel_prom","edu_nivel_min",
              "prop_alfabetos","prop_asiste_edu","alfabeto_jefe"])

# ── VIVIENDA ────────────────────────────────────────────────────
df_cl["idx_vivienda"] = construir_indice(df_cl,
    cols=["material_muros","estado_muros","fuente_agua","sist_excretas",
          "tiene_electricid","hacinamiento_idx","n_dormitorios"],
    invertir=["tiene_electricid","n_dormitorios"])

# ── COMPOSICIÓN ─────────────────────────────────────────────────
df_cl["idx_composicion"] = construir_indice(df_cl,
    cols=["tam_hogar","prop_menores_15","prop_may_65",
          "hay_menores_15","hay_adulto_mayor","n_personas"],
    invertir=[])

# ── INGRESOS ────────────────────────────────────────────────────
df_cl["idx_ingresos"] = construir_indice(df_cl,
    cols=["ingreso_percapita","ingreso_total_hogar","ingreso_jefe",
          "prop_ocupados","prop_con_contrato","contrato_jefe",
          "ingreso_sueldos","ingreso_indep"],
    invertir=["ingreso_percapita","ingreso_total_hogar","ingreso_jefe",
              "prop_ocupados","prop_con_contrato","contrato_jefe",
              "ingreso_sueldos","ingreso_indep"])

INDICES = ["idx_educacion","idx_vivienda","idx_composicion","idx_ingresos"]

print("Índices construidos:")
print(df_cl[INDICES].describe().round(3))
```

## 4. Estandarización con StandardScaler




```python
# Filtrar filas con NaN en algún índice
mask_ok = df_cl[INDICES].notna().all(axis=1)
X_raw = df_cl.loc[mask_ok, INDICES].copy().reset_index(drop=True)
meta  = df_cl.loc[mask_ok, ["folio","año","pobreza_multi","comuna"]].copy().reset_index(drop=True)

# Estandarizar
ss = StandardScaler()
X_scaled = ss.fit_transform(X_raw)

print(f"Hogares para clustering: {len(X_raw)}")
print()
print("Antes de estandarizar (media / std):")
for col in INDICES:
    print(f"  {col:<20}  media={X_raw[col].mean():.3f}  std={X_raw[col].std():.3f}")

print()
print("Después de estandarizar (media / std):")
X_sc_df = pd.DataFrame(X_scaled, columns=INDICES)
for col in INDICES:
    print(f"  {col:<20}  media={X_sc_df[col].mean():.3f}  std={X_sc_df[col].std():.3f}")
```

## 5. Dendrograma — ¿cuántos clusters naturales existen?


```python
Z = linkage(X_scaled, method="ward", metric="euclidean")

fig, ax = plt.subplots(figsize=(14, 6))

dendrogram(
    Z, ax=ax,
    truncate_mode="lastp", p=40,
    leaf_rotation=90, leaf_font_size=8,
    show_contracted=True,
    color_threshold=0.7 * max(Z[:,2]),
)

umbral_corte = 0.7 * max(Z[:,2])
ax.axhline(umbral_corte, color="red", linestyle="--", lw=1.5,
           label=f"Umbral sugerido ({umbral_corte:.1f})")
ax.set_title("Dendrograma — Clustering Jerárquico Ward\nPobreza Multidimensional Los Ríos",
             fontsize=13, fontweight="bold")
ax.set_xlabel("Hogares (truncado a 40 nodos)", fontsize=11)
ax.set_ylabel("Distancia euclidiana (Ward)", fontsize=11)
ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
print()
print("→ Observa los saltos grandes en el eje Y para decidir el número de clusters.")
```

## 6. Número óptimo de clusters — 3 métricas

| Métrica | Qué mide | Mejor valor |
|---------|----------|-------------|
| **Silhouette** | Qué tan separados y compactos son los clusters | Mayor |
| **Calinski-Harabasz** | Ratio varianza entre-clusters / intra-clusters | Mayor |
| **Davies-Bouldin** | Similitud promedio entre cada cluster y su más parecido | Menor |


```python
rango = range(2, 9)
silhouettes, calinski, davies = [], [], []

for k in rango:
    lbl = fcluster(Z, k, criterion="maxclust")
    silhouettes.append(silhouette_score(X_scaled, lbl))
    calinski.append(calinski_harabasz_score(X_scaled, lbl))
    davies.append(davies_bouldin_score(X_scaled, lbl))

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle("Selección del Número Óptimo de Clusters",
             fontsize=13, fontweight="bold")

configs = [
    (axes[0], silhouettes, "#4e79a7", "Silhouette Score",     "Mayor = mejor", max, silhouettes.index(max(silhouettes))),
    (axes[1], calinski,    "#f28e2b", "Calinski-Harabasz",    "Mayor = mejor", max, calinski.index(max(calinski))),
    (axes[2], davies,      "#e15759", "Davies-Bouldin",       "Menor = mejor", min, davies.index(min(davies))),
]

votos = []
for ax, vals, color, titulo, subtitulo, fn, idx_opt in configs:
    k_opt = list(rango)[idx_opt]
    votos.append(k_opt)
    ax.plot(list(rango), vals, "o-", color=color, lw=2, markersize=7)
    ax.axvline(k_opt, color="red", linestyle="--", lw=1.5, label=f"Óptimo k={k_opt}")
    ax.set_title(f"{titulo}\n({subtitulo})", fontsize=10, fontweight="bold")
    ax.set_xlabel("Número de clusters K")
    ax.legend(fontsize=9)
    ax.set_xticks(list(rango))

plt.tight_layout()
plt.show()

k_optimo = Counter(votos).most_common(1)[0][0]
print(f"\nVotos — Silhouette: k={votos[0]} | Calinski: k={votos[1]} | Davies: k={votos[2]}")
print(f"→ K óptimo por votación: {k_optimo}")
```

## 7. Asignar clusters

```python
# Puedes cambiar K manualmente si el dendrograma sugiere otro valor
K = k_optimo   # ← cambiar aquí si prefieres otro K

labels = fcluster(Z, K, criterion="maxclust")
meta["cluster"]  = labels
X_raw["cluster"] = labels

# Guardar en df_cl con el mask correcto
df_cl["cluster"] = np.nan
df_cl.loc[mask_ok, "cluster"] = labels

print(f"K = {K} clusters")
print()
print("Distribución de hogares por cluster:")
vc = meta["cluster"].value_counts().sort_index()
for cl, n in vc.items():
    pct = 100 * n / len(meta)
    barra = "█" * int(pct / 2)
    print(f"  Cluster {cl}: {n:>5} hogares ({pct:.1f}%)  {barra}")
```

## 8. Perfil de cada cluster

```python
perfil = X_raw.groupby("cluster")[INDICES].mean()
perfil.index = [f"Cluster {i}" for i in perfil.index]
perfil.columns = [c.replace("idx_","").capitalize() for c in perfil.columns]

perfil["% Pobres reales"] = (
    meta.groupby("cluster")["pobreza_multi"].mean().values * 100
)
perfil["N hogares"] = meta.groupby("cluster").size().values
perfil["Años (% 2024)"] = (
    meta.groupby("cluster")["año"].apply(lambda x: (x==2024).mean() * 100).values
)

print("=== Perfil de clusters ===")
print(perfil.round(2).to_string())
print()
print("Interpretación del índice: 0 = sin carencia · 1 = máxima carencia")
```

## 9. Heatmap de perfiles

```python
cols_hm  = [c.replace("idx_","").capitalize() for c in INDICES]
datos_hm = perfil[cols_hm]

# Etiquetas del eje Y con % pobres y n hogares
labels_y = [
    f"{idx_cl}  |  {int(perfil.loc[idx_cl,'% Pobres reales'])}% pobres  (n={int(perfil.loc[idx_cl,'N hogares'])})"
    for idx_cl in perfil.index
]

fig, ax = plt.subplots(figsize=(11, K * 1.2))

sns.heatmap(
    datos_hm,
    annot=True, fmt=".2f",
    cmap="RdYlGn_r",
    vmin=0, vmax=1,
    linewidths=1.5,
    linecolor="white",
    ax=ax,
    annot_kws={"size": 13, "weight": "bold"},
    cbar_kws={"label": "Índice de carencia  (0 = sin carencia · 1 = máxima)", "shrink": 0.8},
)

ax.set_title(
    f"Perfil de Carencia por Cluster  (K={K})\nPobreza Multidimensional Los Ríos · CASEN 2017+2024",
    fontsize=13, fontweight="bold", pad=15
)
ax.set_yticklabels(labels_y, rotation=0, fontsize=10)
ax.set_xticklabels(ax.get_xticklabels(), fontsize=11, fontweight="bold")
ax.set_ylabel("")
ax.set_xlabel("")

plt.tight_layout()

plt.show()


```

## 10. Gráfico radar — comparación de perfiles

```python
dims    = cols_hm
N_dims  = len(dims)
angulos = np.linspace(0, 2*np.pi, N_dims, endpoint=False).tolist()
angulos += angulos[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
fig.patch.set_facecolor('#F8F9FA')

for i, (idx_cl, row) in enumerate(perfil[cols_hm].iterrows()):
    vals   = row.tolist() + row.tolist()[:1]
    color  = PALETTE[i % len(PALETTE)]
    n      = int(perfil.loc[idx_cl, "N hogares"])
    pct    = perfil.loc[idx_cl, "% Pobres reales"]
    ax.plot(angulos, vals, "o-", lw=2.5, color=color,
            label=f"{idx_cl}  ({pct:.0f}% pobres · n={n})")
    ax.fill(angulos, vals, alpha=0.1, color=color)

ax.set_xticks(angulos[:-1])
ax.set_xticklabels(dims, fontsize=13, fontweight="bold")
ax.set_ylim(0, 1)
ax.set_yticks([0.25, 0.5, 0.75, 1.0])
ax.set_yticklabels(["0.25","0.50","0.75","1.00"], fontsize=8, color="#999")
ax.set_title(f"Radar de Perfiles (K={K})\nPobreza Multidimensional Los Ríos",
             fontsize=13, fontweight="bold", pad=25)
ax.legend(loc="upper right", bbox_to_anchor=(1.45, 1.15), fontsize=10)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

```

## 11. Distribución de clusters por año

```python
tabla_año = meta.groupby(["año","cluster"]).size().unstack(fill_value=0)
tabla_año.columns = [f"Cluster {c}" for c in tabla_año.columns]
tabla_pct_año = tabla_año.div(tabla_año.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(9, 5))
fig.suptitle("Evolución de Clusters 2017 → 2024\nLos Ríos · CASEN",
             fontsize=13, fontweight="bold")

tabla_pct_año.T.plot(kind="bar", ax=ax,
                     color=PALETTE[:K], edgecolor="white", width=0.6, alpha=0.9)
ax.set_xlabel("Cluster"); ax.set_ylabel("% hogares")
ax.set_title("% de hogares por cluster y año", fontsize=11)
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
ax.legend(title="Año", fontsize=9)
ax.set_ylim(0, 60)

plt.tight_layout()

plt.show()
```

## 12. Distribución de clusters por comuna

```python
if meta["comuna"].notna().sum() > 0:
    tabla_com = meta.groupby(["comuna","cluster"]).size().unstack(fill_value=0)
    tabla_com.columns = [f"Cluster {c}" for c in tabla_com.columns]
    tabla_pct_com = tabla_com.div(tabla_com.sum(axis=1), axis=0) * 100

    # Ordenar por el cluster más pobre (el de mayor % pobres reales)
    cl_mas_pobre = perfil["% Pobres reales"].idxmax().replace("Cluster ","")
    col_ord = f"Cluster {cl_mas_pobre}"
    if col_ord in tabla_pct_com.columns:
        tabla_pct_com = tabla_pct_com.sort_values(col_ord, ascending=False)

    fig, ax = plt.subplots(figsize=(13, 6))
    tabla_pct_com.plot(kind="bar", ax=ax, stacked=True,
                      color=PALETTE[:K], edgecolor="white", width=0.75, alpha=0.9)
    ax.set_xlabel("Comuna", fontsize=11)
    ax.set_ylabel("% de hogares", fontsize=11)
    ax.set_title(f"Composición de Clusters por Comuna (K={K})\nLos Ríos · CASEN 2017+2024",
                 fontsize=13, fontweight="bold")
    ax.set_xticklabels(ax.get_xticklabels(), rotation=30, ha="right", fontsize=10)
    ax.legend(title="Cluster", fontsize=10, bbox_to_anchor=(1.01,1), loc="upper left")
    ax.set_ylim(0, 100)

    plt.tight_layout()
    plt.show()
else:
    print("⚠️  Sin datos de comuna disponibles")
```

## 13. Índices de carencia promedio por comuna

```python
if meta["comuna"].notna().sum() > 0:
    df_cl_com = df_cl[mask_ok].copy()
    df_cl_com["cluster"] = labels

    idx_comuna = df_cl_com.groupby("comuna")[INDICES].mean()
    idx_comuna.columns = [c.replace("idx_","").capitalize() for c in idx_comuna.columns]
    idx_comuna = idx_comuna.sort_values("Ingresos", ascending=False)

    fig, ax = plt.subplots(figsize=(13, 6))
    idx_comuna.plot(kind="bar", ax=ax,
                   color=[COLS_DIM.get(c,"#aaa") for c in idx_comuna.columns],
                   edgecolor="white", width=0.75, alpha=0.9)
    ax.set_xlabel("Comuna", fontsize=11)
    ax.set_ylabel("Índice de carencia promedio (0-1)", fontsize=11)
    ax.set_title("Carencia por Dimensión IPM y Comuna\nLos Ríos · CASEN 2017+2024",
                 fontsize=13, fontweight="bold")
    ax.set_xticklabels(ax.get_xticklabels(), rotation=30, ha="right", fontsize=10)
    ax.legend(title="Dimensión", fontsize=10)
    ax.set_ylim(0, 1)

    plt.tight_layout()
    plt.show()
```

## 14. Guardar resultados para el dashboard

```python
import joblib

import os
os.makedirs("modelos_ml", exist_ok=True)

joblib.dump(K,       "modelos_ml/k_clusters.pkl")
joblib.dump(INDICES, "modelos_ml/indices_cluster.pkl")
df_cl.to_csv("modelos_ml/df_cl.csv", index=False)

print("✅ Guardado:")
print(f"   k_clusters.pkl      → K={K}")
print(f"   indices_cluster.pkl → {INDICES}")
print(f"   df_cl.csv           → {df_cl.shape}")
print(f"   clusters asignados: {df_cl['cluster'].notna().sum()}/{len(df_cl)}")
```
