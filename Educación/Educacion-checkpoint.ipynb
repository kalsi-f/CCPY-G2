```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("casen_losrios_2017_2024.csv")
df_jefes = df[df["jefe_hogar"] == 1]

# agrupa niveles
def agrupar_nivel(codigo):
    if codigo in [1, 2, 3, 4, 5]:
        return "Sin educación formal"
    elif codigo in [6, 7, 8]:
        return "Básica"
    elif codigo in [9, 10, 11]:
        return "Media"
    elif codigo in [12, 13, 14, 15]:
        return "Superior"
    else:
        return None

df_jefes["grupo_educacion"] = df_jefes["nivel_educacion"].apply(agrupar_nivel)

```

```python

# Grafico barrar horizontales
etiquetas_e6a = {
    1: "Nunca asistió",  2: "Sala cuna",
    3: "Jardín infantil", 4: "Kínder",
    5: "Ed. especial",   6: "Primaria antigua",
    7: "Ed. básica",     8: "Humanidades",
    9: "Media C-H",      10: "Media T-P antigua",
    11: "Media T-P",     12: "Técnico superior",
    13: "Universitaria", 14: "Magíster",
    15: "Doctorado"
}

df_2017 = df_jefes[df_jefes["año"] == 2017]
df_2024 = df_jefes[df_jefes["año"] == 2024]

# normalizar
conteo_2017 = df_2017["nivel_educacion"].value_counts(normalize=True) * 100
conteo_2024 = df_2024["nivel_educacion"].value_counts(normalize=True) * 100

# todo en un dataframe
comparacion = pd.DataFrame({
    "2017": conteo_2017,
    "2024": conteo_2024
}).fillna(0)
comparacion["total"] = comparacion["2017"] + comparacion["2024"]
comparacion = comparacion.sort_values("total", ascending=True)  # ascending=True porque barh invierte el orden
comparacion = comparacion.drop(columns=["total"])

comparacion.index = comparacion.index.map(etiquetas_e6a)

fig, ax = plt.subplots(figsize=(11, 7))
x = range(len(comparacion))
alto = 0.35

bars_2017 = ax.barh(
    [i + alto/2 for i in x], comparacion["2017"],
    height=alto, color="#378ADD", label="2017"
)
bars_2024 = ax.barh(
    [i - alto/2 for i in x], comparacion["2024"],
    height=alto, color="#1D9E75", label="2024"
)

ax.set_yticks(list(x))
ax.set_yticklabels(comparacion.index, fontsize=9)
ax.set_xlabel("Porcentaje (%)")
ax.set_xlim(0, 100)
ax.set_title("Nivel educacional de jefes de hogar\n2017 vs 2024 — Los Ríos", fontsize=13)
ax.legend()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()
```

```python

# Grafico 2 Torta
grupos_2017 = df_2017["grupo_educacion"].value_counts()
grupos_2024 = df_2024["grupo_educacion"].value_counts()

# define orden-color
grupos = ["Sin educación formal", "Básica", "Media", "Superior"]
colores = ["#E24B4A", "#378ADD", "#1D9E75", "#F0992B"]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 6))

for ax, grupos_data, año in [(ax1, grupos_2017, "2017"), (ax2, grupos_2024, "2024")]:
    valores = [grupos_data.get(g, 0) for g in grupos]
    wedges, texts, autotexts = ax.pie(
        valores,
        labels=grupos,
        autopct="%1.1f%%",
        colors=colores,
        startangle=90,
        wedgeprops={"edgecolor": "white", "linewidth": 2}
    )
    for text in texts:
        text.set_fontsize(10)
    for autotext in autotexts:
        autotext.set_fontsize(9)
        autotext.set_color("white")
        autotext.set_fontweight("bold")
    ax.set_title(f"Jefes de hogar {año}", fontsize=12)

plt.suptitle("Distribución por grupo educacional — Los Ríos", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

```
